```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv

load_dotenv()
```

```python
model = ChatGroq(model="llama-3.1-8b-instant")
```

```python
class BlogState(TypedDict):
    
    title: str
    outline: str
    content: str
    evaluate: int
```

```python
graph = StateGraph(BlogState)
```

```python
def create_outline(state: BlogState) -> BlogState:
    title = state['title']
    prompt = f"Generate an detailed outline for a blog on the topic - {title}"
    outline = model.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']
    prompt = f"Write a detailed blog on the title - {title} using the following outline \n {outline}"
    content = model.invoke(prompt).content
    state['content'] = content
    return state
```

```python
def evaluate_blog(state: BlogState) -> BlogState:
    outline = state['outline']
    content = state['content']
    
    prompt = f"On a scale of 1 to 10, rate the following **blog content** based on how well it executes the provided **outline**. **Return only the number, with no additional text, explanation, or punctuation.** \n\nOutline: '{outline}' \n\nBlog Content: '{content}'"
    
    evaluate = model.invoke(prompt)
    print(evaluate)
    state['evaluate'] = evaluate
    return state
```

```python
graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)
graph.add_node("evaluate_blog", evaluate_blog)
```

```python
graph.add_edge(START, "create_outline")
graph.add_edge("create_outline", "create_blog")
graph.add_edge("create_blog", "evaluate_blog")
graph.add_edge("evaluate_blog", END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {"title": "Rise of AI in India"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
